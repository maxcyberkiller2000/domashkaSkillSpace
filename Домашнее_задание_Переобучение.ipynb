```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
import matplotlib.pyplot as plt

# Параметры
VOCAB_SIZE = 10000
MAX_LENGTH = 300
EMBEDDING_DIM = 128
BATCH_SIZE = 128
EPOCHS = 8

# Загрузка и подготовка данных
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=VOCAB_SIZE)

# Анализ данных
print(f"Размер тренировочной выборки: {len(X_train)}")
print(f"Размер тестовой выборки: {len(X_test)}")
print(f"Пример метки: {y_train[0]}")
print(f"Длина первого отзыва: {len(X_train[0])}")

# Паддинг последовательностей
X_train = pad_sequences(X_train, maxlen=MAX_LENGTH)
X_test = pad_sequences(X_test, maxlen=MAX_LENGTH)

# Разделение на тренировочную и валидационную выборки
VAL_SPLIT = 0.2
indices = np.arange(X_train.shape[0])
np.random.shuffle(indices)
X_train = X_train[indices]
y_train = y_train[indices]

num_val_samples = int(VAL_SPLIT * X_train.shape[0])
X_val = X_train[:num_val_samples]
y_val = y_train[:num_val_samples]
X_train_partial = X_train[num_val_samples:]
y_train_partial = y_train[num_val_samples:]

print(f"Тренировочная выборка: {X_train_partial.shape[0]}")
print(f"Валидационная выборка: {X_val.shape[0]}")
print(f"Контрольная выборка: {X_test.shape[0]}")

# Создание модели
def create_model():
    model = models.Sequential([
        layers.Embedding(VOCAB_SIZE, EMBEDDING_DIM, input_length=MAX_LENGTH),

        # Сверточные слои для извлечения features
        layers.Conv1D(128, 5, activation='relu'),
        layers.GlobalMaxPooling1D(),

        # Полносвязные слои
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.5),
        layers.Dense(64, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer='adam',
        loss='binary_crossentropy',
        metrics=['accuracy', 'precision', 'recall']
    )

    return model

# Создание и вывод структуры модели
model = create_model()
print("Архитектура модели:")
model.summary()

# Callbacks для улучшения обучения
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_accuracy',
        patience=3,
        restore_best_weights=True
    ),
    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=2,
        min_lr=0.0001
    )
]

# Обучение модели
print("Начало обучения...")
history = model.fit(
    X_train_partial, y_train_partial,
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    validation_data=(X_val, y_val),
    callbacks=callbacks,
    verbose=1
)

# Оценка на валидационной выборке
print("\nОценка на валидационной выборке...")
val_loss, val_accuracy, val_precision, val_recall = model.evaluate(X_val, y_val, verbose=0)

# Оценка на контрольной выборке
print("Оценка на контрольной выборке...")
test_loss, test_accuracy, test_precision, test_recall = model.evaluate(X_test, y_test, verbose=0)

# Вывод результатов
print("\n" + "="*50)
print("РЕЗУЛЬТАТЫ:")
print("="*50)
print(f"Валидационная выборка - Точность: {val_accuracy:.4f} ({val_accuracy*100:.2f}%)")
print(f"Контрольная выборка - Точность: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"Precision на контроле: {test_precision:.4f}")
print(f"Recall на контроле: {test_recall:.4f}")

# Проверка условий задания
print("\n" + "="*50)
print("="*50)

val_success = val_accuracy > 0.90
test_success = test_accuracy > 0.88

if val_accuracy > 0.95 and test_accuracy > 0.93:
    score = 4
elif val_accuracy > 0.90 and test_accuracy > 0.88:
    score = 3
else:
    score = 0

print(f"✓ Точность на валидации > 90%: {val_accuracy*100:.2f}% {'✅' if val_success else '❌'}")
print(f"✓ Точность на контроле > 88%: {test_accuracy*100:.2f}% {'✅' if test_success else '❌'}")

# Визуализация процесса обучения
plt.figure(figsize=(15, 5))

# График точности
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Точность модели')
plt.xlabel('Эпоха')
plt.ylabel('Точность')
plt.legend()
plt.grid(True)

# График потерь
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Потери модели')
plt.xlabel('Эпоха')
plt.ylabel('Потери')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

# Функция для предсказания на новых отзывах
def predict_sentiment(text_sequence):
    """Предсказание тональности отзыва"""
    if len(text_sequence) == 0:
        return "Ошибка: пустая последовательность"

    # Паддинг и преобразование в правильный формат
    text_padded = pad_sequences([text_sequence], maxlen=MAX_LENGTH)

    # Предсказание
    prediction = model.predict(text_padded, verbose=0)[0][0]

    sentiment = "ПОЛОЖИТЕЛЬНЫЙ" if prediction > 0.5 else "ОТРИЦАТЕЛЬНЫЙ"
    confidence = prediction if prediction > 0.5 else 1 - prediction

    return f"Отзыв: {sentiment} (уверенность: {confidence:.2%})"

# Демонстрация работы на примерах из тестовой выборки
print("\n" + "="*50)
print("ДЕМОНСТРАЦИЯ РАБОТЫ МОДЕЛИ:")
print("="*50)

# Словарь для обратного преобразования индексов в слова
word_index = imdb.get_word_index()
reverse_word_index = {value: key for key, value in word_index.items()}

def decode_review(encoded_review):
    return ' '.join([reverse_word_index.get(i - 3, '?') for i in encoded_review if i >= 3])

# Тестирование на нескольких примерах
for i in range(3):
    sample_idx = np.random.randint(0, len(X_test))
    sample_text = X_test[sample_idx]
    true_label = y_test[sample_idx]

    prediction = model.predict(pad_sequences([sample_text], maxlen=MAX_LENGTH), verbose=0)[0][0]
    predicted_label = 1 if prediction > 0.5 else 0

    print(f"\nПример {i+1}:")
    print(f"Истинная метка: {'ПОЛОЖИТЕЛЬНЫЙ' if true_label == 1 else 'ОТРИЦАТЕЛЬНЫY'}")
    print(f"Предсказание: {'ПОЛОЖИТЕЛЬНЫЙ' if predicted_label == 1 else 'ОТРИЦАТЕЛЬНЫЙ'} (вероятность: {prediction:.4f})")
    print(f"Результат: {'✅ ВЕРНО' if predicted_label == true_label else '❌ ОШИБКА'}")
```
