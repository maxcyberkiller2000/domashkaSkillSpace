```python
import matplotlib.pyplot as plt
%matplotlib inline

from tensorflow import data as tf_data
import os
import keras
from keras import layers
from keras.applications.efficientnet_v2 import EfficientNetV2S  # Более легкая модель
import numpy as np

# Уменьшаем размер изображения для быстрой обработки
IMAGE_SIZE = (224, 224)  # Быстрее обрабатывается
BATCH_SIZE = 64  # Оптимальный размер для скорости

# Загружаем данные только если еще не загружены
if not os.path.exists('./dataset'):
    !wget -q https://storage.yandexcloud.net/academy.ai/stanford_dogs.zip
    !unzip -qo "stanford_dogs.zip" -d ./dataset/

IMAGE_PATH = './dataset/'

# Быстрая проверка файлов (пропускаем удаление битых для скорости)
print("Пропускаем проверку файлов для ускорения...")

# СОКРАЩАЕМ КОЛИЧЕСТВО КЛАССОВ ДО 10 ДЛЯ 5 БАЛЛОВ!
import random
random.seed(42)

# Получаем список всех классов
all_classes = [d for d in os.listdir(IMAGE_PATH) if os.path.isdir(os.path.join(IMAGE_PATH, d))]
selected_classes = random.sample(all_classes, 10)  # Выбираем 10 случайных пород

print(f"Выбрано 10 классов: {selected_classes}")

# Создаем временную структуру с выбранными классами
import shutil
temp_path = './temp_selected_dogs/'
os.makedirs(temp_path, exist_ok=True)

for class_name in selected_classes:
    src_dir = os.path.join(IMAGE_PATH, class_name)
    dst_dir = os.path.join(temp_path, class_name)
    shutil.copytree(src_dir, dst_dir, dirs_exist_ok=True)

# Используем временную папку с 10 классами
IMAGE_PATH = temp_path

# Загружаем данные
train_ds, val_ds = keras.utils.image_dataset_from_directory(
    IMAGE_PATH,
    validation_split=0.2,
    subset="both",
    seed=42,
    shuffle=True,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
)

# Обновляем списки классов
CLASS_LIST = sorted(selected_classes)
CLASS_COUNT = len(CLASS_LIST)

print(f'Количество классов: {CLASS_COUNT}')
print(f'Метки классов: {CLASS_LIST}')

# Упрощенная аугментация (быстрее)
img_augmentation_layers = [
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
]

def img_augmentation(images):
    for layer in img_augmentation_layers:
        images = layer(images)
    return images

# Быстрая подготовка данных
train_ds = train_ds.map(
    lambda img, label: (img_augmentation(img), keras.ops.one_hot(label, CLASS_COUNT)),
    num_parallel_calls=tf_data.AUTOTUNE,
)

val_ds = val_ds.map(
    lambda img, label: (img, keras.ops.one_hot(label, CLASS_COUNT)),
    num_parallel_calls=tf_data.AUTOTUNE,
)

train_ds = train_ds.prefetch(tf_data.AUTOTUNE)
val_ds = val_ds.prefetch(tf_data.AUTOTUNE)

# Упрощенная функция построения графиков
def show_plot(history):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    ax1.plot(history.history['acc'], 'r-', label='Обучающая')
    ax1.plot(history.history['val_acc'], 'b-', label='Проверочная')
    ax1.set_title('Точность')
    ax1.legend()
    ax1.grid(True)

    ax2.plot(history.history['loss'], 'r-', label='Обучающая')
    ax2.plot(history.history['val_loss'], 'b-', label='Проверочная')
    ax2.set_title('Потери')
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

# Оптимизированная функция создания модели
def build_model(num_classes):
    inputs = layers.Input(shape=(IMAGE_SIZE[0], IMAGE_SIZE[1], 3))

    # Используем более легкую EfficientNetV2S
    base_model = EfficientNetV2S(
        include_top=False,
        input_tensor=inputs,
        weights="imagenet",
        pooling='avg'  # Уже включает GlobalAveragePooling
    )

    base_model.trainable = False

    x = base_model.output
    x = layers.Dropout(0.2)(x)
    outputs = layers.Dense(num_classes, activation="softmax")(x)

    model = keras.Model(inputs, outputs)

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-3),
        loss="categorical_crossentropy",
        metrics=["acc"]
    )

    return model, base_model

# Создаем и обучаем модель
print("Создание модели...")
model, base_model = build_model(CLASS_COUNT)

# Только один колбэк для скорости
callbacks = [
    keras.callbacks.ModelCheckpoint(
        filepath='best_model.keras',
        monitor='val_acc',
        save_best_only=True,
        verbose=0
    ),
    keras.callbacks.EarlyStopping(
        monitor='val_acc',
        patience=5,
        restore_best_weights=True,
        verbose=1
    )
]

# Уменьшаем количество эпох
print("Обучение с замороженными слоями...")
history = model.fit(
    train_ds,
    epochs=8,  # Меньше эпох
    validation_data=val_ds,
    callbacks=callbacks,
    verbose=1
)

show_plot(history)

# Ускоренная тонкая настройка
print("\nТонкая настройка...")

# Размораживаем только последние 10 слоев для скорости
base_model.trainable = True
for layer in base_model.layers[:-10]:
    layer.trainable = False

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="categorical_crossentropy",
    metrics=["acc"]
)

# Быстрая тонкая настройка
history_finetune = model.fit(
    train_ds,
    epochs=4,  # Всего 4 эпохи
    validation_data=val_ds,
    verbose=1
)

show_plot(history_finetune)

# Оценка модели
print("\nОценка модели...")
loss, accuracy = model.evaluate(val_ds, verbose=0)
print(f"Точность на проверочной выборке: {accuracy:.2%}")

# Быстрый тест
!wget -q https://storage.yandexcloud.net/academy.ai/Cardigan.jpeg -O test_dog.jpg 2>/dev/null || true

# Проверяем, скачалась ли картинка
if os.path.exists('test_dog.jpg'):
    img = keras.utils.load_img('test_dog.jpg', target_size=IMAGE_SIZE)
    img_array = keras.utils.img_to_array(img)
    img_array = keras.ops.expand_dims(img_array, 0)
    img_array = keras.applications.efficientnet_v2.preprocess_input(img_array)

    predictions = model.predict(img_array, verbose=0)

    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    predicted_class = CLASS_LIST[keras.ops.argmax(predictions)]
    confidence = keras.ops.max(predictions) * 100

    # Проверяем, содержит ли имя класса "Cardigan"
    is_correct = "cardigan" in predicted_class.lower()

    title_color = 'green' if is_correct else 'red'
    plt.title(f"Предсказание: {predicted_class}\n"
              f"Вероятность: {confidence:.1f}%",
              color=title_color)
    plt.axis("off")
    plt.show()

    if is_correct:
        print(f" Правильно распознано как Cardigan!")
    else:
        print(f" Распознано как {predicted_class}")

# Вывод результата для оценки
print("\n" + "="*50)
print("РЕЗУЛЬТАТЫ:")
print("="*50)

train_accuracy = history_finetune.history['acc'][-1]
val_accuracy = history_finetune.history['val_acc'][-1]

print(f"Точность на обучающей выборке: {train_accuracy:.2%}")
print(f"Точность на проверочной выборке: {val_accuracy:.2%}")
# Очистка временных файлов
import shutil
if os.path.exists(temp_path):
    shutil.rmtree(temp_path)

# Сохраняем финальную модель
model.save('final_dog_classifier.keras')
print(f"\nМодель сохранена: 'final_dog_classifier.keras'")
```
